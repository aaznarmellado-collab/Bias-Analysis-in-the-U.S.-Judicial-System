```python
df = pd.read_csv(r"..\00_Data\00_Processed\df_eda.csv")
```

```python
df = convert_to_datetime(df)
df = convert_categorical(df)
```

```python
eda(df)
```

```python
# We assume a decile score of 5 or more is a positive prediction.

df['prediction'] = (df['decile_score'] >= 5).astype(int)
check_bias(df)
```

```python
lista_aux = ["c_charge_degree", "juv_priors_count", "adult_priors_count", "sex", "age_cat", "maritalstatus", "race", "is_recid", 'two_year_recid', 'decile_score', 'prediction']
df_aux = df[lista_aux]

df_aux['priors_count_cat'] = pd.cut(
    df_aux['adult_priors_count'],
    bins=[-1, 1, 4, np.inf],
    labels=['Bajo', 'Medio', 'Alto']
)

df_aux["juv_priors_flag"] = (df_aux["juv_priors_count"] > 0).map({True: "Sí", False: "No"})
```

```python
# Subsets combining different relevant variables. 

priors = df_aux.priors_count_cat.cat.categories.tolist()
juv = df_aux.juv_priors_flag.unique().tolist()

for p, j in product(priors, juv):

    subset = df_aux[
        (df_aux['priors_count_cat'] == p) &
        (df_aux['juv_priors_flag'] == j)
    ]

    print("\n" + "=" * 50)
    print(f"GROUP: {p} | {j} | n={len(subset)}")
    print("=" * 50)

    check_bias(subset)
```

```python
charge = df_aux.c_charge_degree.unique().tolist()

for c in charge:

    subset = df_aux[
        df_aux['c_charge_degree'] == c
    ]

    # Skip empty subsets
    if len(subset) == 0:
        continue

    print("\n" + "=" * 50)
    print(f"GRUPO: {c} n={len(subset)}")
    print("=" * 50)

    check_bias(subset)
```

```python
priors = df_aux.priors_count_cat.cat.categories.tolist()
juv = df_aux.juv_priors_flag.unique().tolist()
sex = df_aux.sex.unique().tolist()
age_cat = df_aux.age_cat.cat.categories.tolist()
maritalstatus = df_aux.maritalstatus.unique().tolist()

for p, j, s, a, m in product(priors, juv, sex, age_cat, maritalstatus):

    subset = df_aux[
        (df_aux['priors_count_cat'] == p) &
        (df_aux['juv_priors_flag'] == j) &
        (df_aux['sex'] == s) &
        (df_aux['age_cat'] == a) &
        (df_aux['maritalstatus'] == m)
    ]

    # evitar grupos vacíos
    if len(subset) == 0:
        continue

    print("\n" + "=" * 50)
    print(f"GRUPO: {p} | {j} | {s} | {a} | {m} |n={len(subset)}")
    print("=" * 50)

    check_bias(subset)
```

### COMPAS confusion matrix for reference

```python
cm_total = confusion_matrix(df_aux['two_year_recid'], df_aux['prediction'], labels=[0,1])
cm_total_all = confusion_matrix(df_aux['two_year_recid'], df_aux['prediction'], labels=[0,1], normalize = 'all')
cm_total_true = confusion_matrix(df_aux['two_year_recid'], df_aux['prediction'], labels=[0,1], normalize = 'true')
```

```python
disp = ConfusionMatrixDisplay(cm_total)
disp.plot()
disp = ConfusionMatrixDisplay(cm_total_all)
disp.plot()
disp = ConfusionMatrixDisplay(cm_total_true)
disp.plot()

plt.show()
```
